# **Defining Functions for Different Neural Network Operations**

Function to Initalize parameters

````python
def init_params():

    W1 = np.ones((784, 128))*0.1
    b1 = np.zeros((128, 1))

    W2 = np.ones((128, 64))*0.1
    b2 = np.zeros((64, 1))

    W3 = np.ones((64, 10))*0.1
    b3 = np.zeros((10, 1))

    return W1, b1, W2, b2, W3, b3
````

Activation Functions

````python
def relu(Z):
    # Returns max(0, Z)
    return np.maximum(0, Z)


def softmax(Z):
    exp_Z = np.exp(Z - np.max(Z, axis=0, keepdims=True))
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)
````

Linear Forward Function / Perceptron

````python
def linear_forward(A_prev, W, b):
    # Linear transformation
    Z = np.dot(W.T, A_prev) + b
    return Z
````

Function for Forward Propagation

````python
def forward_propagation(X, W1, b1, W2, b2, W3, b3):

    # Layer 1
    Z1 = linear_forward(X, W1, b1)
    A1 = relu(Z1)

    # Layer 2
    Z2 = linear_forward(A1, W2, b2)
    A2 = relu(Z2)

    # Output layer
    Z3 = linear_forward(A2, W3, b3)
    A3 = softmax(Z3)

    return Z1, A1, Z2, A2, Z3, A3
````

## Forward Propagation

The forward propagation process passes the input through each layer of the neural network to generate the final prediction.

```text

└─────────────────────┘
        │
        ▼

        ▼
    Softmax(Z₃)

        │

    Final Prediction

Compute Loss

````python
def calculate_loss(Y, Y_hat):
    m = Y.shape[0]

    loss = -np.sum(Y * np.log(Y_hat + 1e-8)) / m

    return loss
````

````python
def one_hot(Y):
    one_hot_Y = np.zeros((Y.size, Y.max() + 1))
    one_hot_Y[np.arange(Y.size), Y] = 1
    one_hot_Y = one_hot_Y.T
    return one_hot_Y
````

````python
one_hot(y_train)
````

Back Propagation

````python
def backpropagation(X, Y, W1, b1, W2, b2, W3, b3, A1, A2, A3):
    Y = one_hot(Y)

    m = X.shape[1]

    # Output layer gradient
    dZ3 = A3 - Y
    dW3 = np.dot(A2, dZ3.T) / m
    db3 = np.sum(dZ3) / m

    # Hidden layer 2 gradient
    dA2 = np.dot(W3, dZ3)
    dZ2 = dA2 * (A2 > 0)
    dW2 = np.dot(A1, dZ2.T) / m
    db2 = np.sum(dZ2) / m

    # Hidden layer 1 gradient
    dA1 = np.dot(W2, dZ2)
    dZ1 = dA1 * (A1 > 0)
    dW1 = np.dot(X, dZ1.T) / m
    db1 = np.sum(dZ1) / m

    return dW1, db1, dW2, db2, dW3, db3
````

Update Function for Weights and Bias

````python
def update_parameters(W1, b1, W2, b2, W3, b3, dW1, db1, dW2, db2, dW3, db3, learning_rate):
  #layer 1
  W1 = W1 - learning_rate * dW1
  b1 = b1 - learning_rate * db1

  #layer 2
  W2 = W2 - learning_rate * dW2
  b2 = b2 - learning_rate * db2

  #layer 3
  W3 = W3 - learning_rate * dW3
  b3 = b3 - learning_rate * db3

  return W1, b1, W2, b2, W3, b3
````

Function to get Prediction


````python
def get_predictions(A3):
    return np.argmax(A3, 0)
````

Function to get accuracy

````python
def get_accuracy(predictions, Y):
    print(predictions, Y)
    return np.sum(predictions == Y) / Y.size
````

Gradient Descent Function

````python
def fit(X, Y, learning_rate, iterations):
    W1, b1, W2, b2, W3, b3 = init_params()
    for i in range(iterations):
        Z1, A1, Z2, A2, Z3, A3 = forward_propagation(X, W1, b1, W2, b2, W3, b3)
        dW1, db1, dW2, db2, dW3, db3 = backpropagation(X, Y, W1, b1, W2, b2, W3, b3, A1, A2, A3)
        W1, b1, W2, b2, W3, b3 = update_parameters(W1, b1, W2, b2, W3, b3, dW1, db1, dW2, db2, dW3, db3, learning_rate)
        if i % 10 == 0:
            print("Iteration: ", i)
            predictions = get_predictions(A3)
            print(get_accuracy(predictions, Y))
    return W1, b1, W2, b2, W3, b3
````

````python
W1, b1, W2, b2, W3, b3 = fit(x_train, y_train, 0.1, 100)
````

````python
def make_predictions(X, W1, b1, W2, b2, W3, b3):
    _, _, _, _, _, A3 = forward_propagation(X, W1, b1, W2, b2, W3, b3)
    predictions = get_predictions(A3)
    return predictions

def test_prediction(index, W1, b1, W2, b2, W3, b3):
    current_image = x_train[:, index, None]
    prediction = make_predictions(x_train[:, index, None], W1, b1, W2, b2, W3, b3)
    label = y_train[index]
    print("Prediction: ", prediction)
    print("Label: ", label)

    current_image = current_image.reshape((28, 28)) * 255
    plt.gray()
    plt.imshow(current_image, interpolation='nearest')
    plt.show()
````
