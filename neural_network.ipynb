# **Defining Functions for Different Neural Network Operations**

Function to Initalize parameters

````python
def init_parameters(layers_dims):
    """
    Initialize weights and biases for each layer of the neural network.

    Parameters:
        layers_dims (list): Number of neurons in each layer.
                           Example: [784, 128, 64, 10]

    Returns:
        parameters (dict): Dictionary containing weights and biases
                           for all layers.
    """

    # Store the parameters of each layer in a dictionary
    parameters = {}

    # Total number of layers, including the input layer
    L = len(layers_dims)

    # Initialize weights and biases for each layer
    for l in range(1, L):

        # Weight matrix connecting layer (l-1) to layer l
        parameters['W' + str(l)] = np.ones((layers_dims[l - 1], layers_dims[l]))

        # Bias vector for layer l
        parameters['b' + str(l)] = np.zeros((layers_dims[l], 1))

    return parameters
````

Activation Functions

````python
def relu(Z):
    # Returns max(0, Z)
    return np.maximum(0, Z)


def softmax(Z):
    # Convert logits into probabilities
    exp_Z = np.exp(Z - np.max(Z, axis=0, keepdims=True))
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)
````

Linear Forward Function / Perceptron

````python
def linear_forward(A_prev, W, b):
    # Linear transformation
    Z = np.dot(W.T, A_prev) + b
    return Z
````

Function for Forward Propagation

````python
def forward_propagation(X, parameters):

    # Layer 1
    Z1 = linear_forward(X, parameters['W1'], parameters['b1'])
    A1 = relu(Z1)

    # Layer 2
    Z2 = linear_forward(A1, parameters['W2'], parameters['b2'])
    A2 = relu(Z2)

    # Output layer
    Z3 = linear_forward(A2, parameters['W3'], parameters['b3'])
    A3 = softmax(Z3)

    return A3
````

Back Propagation

````python
def backpropagation(X, Y, parameters, cache):
    # Retrieve forward propagation values
    A1, A2, A3 = cache

    W1 = parameters['W1']
    W2 = parameters['W2']
    W3 = parameters['W3']

    m = X.shape[1]

    # Output layer gradient
    dZ3 = A3 - Y
    dW3 = np.dot(A2, dZ3.T) / m
    db3 = np.sum(dZ3, axis=1, keepdims=True) / m

    # Hidden layer 2 gradient
    dA2 = np.dot(W3, dZ3)
    dZ2 = dA2 * (A2 > 0)
    dW2 = np.dot(A1, dZ2.T) / m
    db2 = np.sum(dZ2, axis=1, keepdims=True) / m

    # Hidden layer 1 gradient
    dA1 = np.dot(W2, dZ2)
    dZ1 = dA1 * (A1 > 0)
    dW1 = np.dot(X, dZ1.T) / m
    db1 = np.sum(dZ1, axis=1, keepdims=True) / m

    # Store gradients
    grads = {
        'dW1': dW1,
        'db1': db1,
        'dW2': dW2,
        'db2': db2,
        'dW3': dW3,
        'db3': db3
    }

    return grads
````
